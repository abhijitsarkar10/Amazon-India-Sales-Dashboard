# **1. Overall Sales & Profit Performance**

```python
total_sales = df["Total_Sales_INR"].sum()
total_profit = df["Profit_INR"].sum()
total_orders = df["Order_ID"].nunique()
total_units = df["Quantity"].sum()

average_order_value = (total_sales / total_orders if total_orders else 0 )

profit_margin = (total_profit / total_sales * 100 if total_sales else 0)

print("========== AMAZON INDIA KPIs ==========")
print(f"Total Sales       : ₹{total_sales:,.2f}")
print(f"Total Profit      : ₹{total_profit:,.2f}")
print(f"Total Orders      : {total_orders:,}")
print(f"Total Units Sold  : {total_units:,}")
print(f"Average Order Value: ₹{average_order_value:,.2f}")
print(f"Profit Margin     : {profit_margin:.2f}%")
```

```python
monthly = (
df.groupby("Year_Month")
    .agg(
        Sales=("Total_Sales_INR", "sum"),
        Profit=("Profit_INR", "sum"),
        Orders=("Order_ID", "nunique")
    )
    .reset_index()
    .sort_values("Year_Month")
)

display(monthly.head(10))
```

# **The monthly sales and profit trend**

```python
fig = px.line(monthly, x="Year_Month", y=["Sales", "Profit"], title="Monthly Sales & Profit", markers=True)
fig.show()
```

```python
category_summary = (
    df.groupby("Category")
    .agg(
        Sales=("Total_Sales_INR", "sum"),
        Profit=("Profit_INR", "sum"),
        Units_Sold=("Quantity", "sum"),
        Orders=("Order_ID", "nunique")
    )
    .reset_index()
    .sort_values("Sales", ascending=False)
)

display(category_summary.round(2))
```

# **2.Category & Product Performance**

## **Top 10 Products by Sales**

# **3. Order Status & Revenue Loss**

```python
print(df["Order_Status"].value_counts())
```

```python
status_summary = df.groupby("Order_Status").agg(
    Orders=("Order_ID", "nunique"),
    Sales=("Total_Sales_INR", "sum"),
    Profit=("Profit_INR", "sum")
).reset_index()

status_summary["Order_Percentage"] = status_summary["Orders"] / total_orders * 100

display(status_summary.round(2))
```

```python
status = df["Order_Status"].str.strip().str.lower()

returned = df.loc[status == "returned", "Order_ID"].nunique()
cancelled = df.loc[status == "cancelled", "Order_ID"].nunique()

return_rate = returned / total_orders * 100
cancel_rate = cancelled / total_orders * 100

print(f"Returned Orders: {returned}")
print(f"Cancelled Orders: {cancelled}")
print(f"Return Rate: {return_rate:.2f}%")
print(f"Cancellation Rate: {cancel_rate:.2f}%")
```

## **Order Status Overview**

```python
fig = px.bar(status_summary, x="Order_Status", y="Orders",
             color="Order_Status", title="Order Status Overview",
             text="Orders")
fig.update_layout(template="plotly_white", showlegend=False)
fig.show()
```

## **Category-wise Returns & Cancellations**

```python
category_status = df.assign(Status_Clean=status).groupby(
    ["Category", "Status_Clean"]
).agg(Orders=("Order_ID", "nunique")).reset_index()

loss_status = category_status[
    category_status["Status_Clean"].isin(["returned", "cancelled"]
)]

fig = px.bar(loss_status, x="Category", y="Orders",
             color="Status_Clean", barmode="group",
             title="Category-wise Returns & Cancellations",
             text_auto=True)
fig.show()
```

# **4. Payment, Fulfillment & Geographic Performance**

```python
payment_summary = (
    df.groupby("Payment_Method")
    .agg(
        Sales=("Total_Sales_INR", "sum"),
        Profit=("Profit_INR", "sum"),
        Orders=("Order_ID", "nunique")
    )
    .reset_index()
    .sort_values("Sales", ascending=False)
)

display(payment_summary.round(2))
```

## **Sales by Payment Method**

```python
fig= px.bar(payment_summary, x="Sales", y="Payment_Method", title="Sales by Payment Method", orientation="h",  text_auto=".2s")
fig.show()
```

## **Profit by Payment Method**

```python
fig = px.bar(payment_summary, x="Profit", y="Payment_Method", title="Profit by Payment Method", orientation="h",  text_auto=".2s")
fig.show()
```

```python
fulfillment_summary = (
    df.groupby("Fulfillment")
    .agg(
        Sales=("Total_Sales_INR", "sum"),
        Profit=("Profit_INR", "sum"),
        Orders=("Order_ID", "nunique"),
        Units_Sold=("Quantity", "sum")
    )
    .reset_index()
    .sort_values("Sales", ascending=False)
)

display(fulfillment_summary.round(2))
```

## **Sales by Fulfilment Method**

```python
fig = px.bar(fulfillment_summary, x="Sales", y="Fulfillment", title="Sales by Fulfilment Method", orientation="h",  text_auto=".2s")
fig.show()
```

## **Profit by Fulfilment Method**

```python
fig = px.bar(fulfillment_summary, x="Profit", y="Fulfillment", title="Profit by Fulfilment Method", orientation="h",  text_auto=".2s")
fig.show()
```

```python
state_summary = (
    df.groupby("Ship_State")
    .agg(
        Sales=("Total_Sales_INR", "sum"),
        Profit=("Profit_INR", "sum"),
        Orders=("Order_ID", "nunique"),
        Units_Sold=("Quantity", "sum")
    )
    .reset_index()
    .sort_values("Sales", ascending=False)
)

display(state_summary.round(2))
```

## **Top 15 State by Sales**

```python
fig = px.bar(state_summary.head(15).sort_values("Sales"), x="Sales", y="Ship_State", title="Top 15 State by Sales", orientation="h",  text_auto=".2s")
fig.show()
```

## **Top 15 State by Profit**

```python
fig= px.bar(state_summary.head(15).sort_values("Profit"), x="Profit", y="Ship_State", title="Top 15 State by Profit", orientation="h",  text_auto=".2s")
fig.show()
```
